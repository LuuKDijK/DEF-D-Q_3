## *Opdracht 6*: Meten en Fitten, Leerdoel

```python
# Zet hier jouw data in een voor jou handig formaat
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as opt
massa_planken = np.array([160,198.8, 272.5, 313.3, 40.8, 79.6,109.8, 184.3, 263.9]) # gram
waarde_tabel = np.array([488,  460, 429, 423, 771, 696, 688, 587, 510 ])

# Errorbars
massa_error = 5  # gram
waarde_error = 20   # waarde

plt.figure(figsize=(10,6))
plt.errorbar(massa_planken, waarde_tabel, xerr=massa_error, yerr=waarde_error, fmt='o', capsize=5, label='Meetpunten met errors')
plt.xlabel('Massa (gram)')
plt.ylabel('Waarde')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Plot hier de meting en fit een lijn door de metingen.
# Maak gebruik van wat je geleerd hebt bij het practicum.
%matplotlib inline

# Fit een lineaire functie y = a*x + b
coefficients = np.polyfit(massa_planken, waarde_tabel, 1)
a, b = coefficients[0], coefficients[1]
fit_line = a * massa_planken + b

# Plot de meetpunten en de fitlijn
plt.figure(figsize=(10, 6))
plt.plot(massa_planken, waarde_tabel, 'o', label='Meetpunten', markersize=8)
plt.plot(massa_planken, fit_line, 'r-', label=f'Fit: y = {a:.2f}x + {b:.2f}')
plt.xlabel('Massa (gram)')
plt.ylabel('Waarde')
plt.legend()
plt.grid(True, alpha=0.3)
plt.title('Kalibratie Meting met Fit')
plt.show()

# Print de fitparameters
print(f"Fitparameters:")
print(f"a (helling) = {a:.4f}")
print(f"b (intercept) = {b:.4f}")
print(f"Vergelijking: y = {a:.4f}*x + {b:.4f}")
```

```python
# Zet hier jouw data in een voor jou handig formaat
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as opt

# Eerste iteratie
massa_planken = np.array([160, 198.8, 272.5, 313.3, 40.8, 79.6, 109.8, 184.3, 263.9])  # gram
waarde_tabel = np.array([488, 460, 429, 423, 771, 696, 688, 587, 510])

# Tweede iteratie
massa_2e_iteratie = np.array([1250, 3750, 2500, 5000, 7500, 8750, 6250])
waarde_2e_iteratie = np.array([234, 109, 140, 52, 35, 30, 40])

# Errorbars
massa_error = 5  # gram
waarde_error = 20    # waarde

# Combineer beide datasets
massa_gecombineerd = np.concatenate([massa_planken, massa_2e_iteratie])
waarde_gecombineerd = np.concatenate([waarde_tabel, waarde_2e_iteratie])

# Fit logaritmische functie over beide iteraties: y = a*log(x) + b
sorted_indices = np.argsort(massa_gecombineerd)
x_sorted = massa_gecombineerd[sorted_indices]
y_sorted = waarde_gecombineerd[sorted_indices]

# Fit: log scale
coeffs = np.polyfit(np.log10(x_sorted), y_sorted, 1)
fit_line = coeffs[0] * np.log10(x_sorted) + coeffs[1]

plt.figure(figsize=(10, 6))
# Eerste iteratie (blauwe kleur)
plt.errorbar(massa_planken, waarde_tabel, xerr=massa_error, yerr=waarde_error, 
             fmt='o', capsize=5, color='blue', label='1e iteratie', markersize=8)

# Tweede iteratie (groene kleur)
plt.errorbar(massa_2e_iteratie, waarde_2e_iteratie, xerr=massa_error, yerr=waarde_error, 
             fmt='s', capsize=5, color='green', label='2e iteratie', markersize=8)

# Één fitlijn over beide iteraties
plt.plot(x_sorted, fit_line, 'r--', linewidth=2, label=f'Fit: y = {coeffs[0]:.2f}*log(x) + {coeffs[1]:.2f}')

plt.xlabel('Massa (gram)')
plt.ylabel('Waarde')
plt.legend()
plt.grid(True, alpha=0.3)
plt.title('Kalibratie Metingen - Iteratie 1 en 2')
plt.xscale('log')  # Logaritmische schaal voor de x-as
plt.show()

print(f"Fitparameters (beide iteraties gecombineerd):")
print(f"a (helling) = {coeffs[0]:.4f}")
print(f"b (intercept) = {coeffs[1]:.4f}")
print(f"Vergelijking: y = {coeffs[0]:.4f}*log(x) + {coeffs[1]:.4f}")
```
